```python
from google.colab import files
uploaded = files.upload()
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# DLT Function
def dlt(src_pts, dst_pts):
    A = []

    for i in range(len(src_pts)):
        x, y = src_pts[i]
        xp, yp = dst_pts[i]

        A.append([-x, -y, -1, 0, 0, 0, xp*x, xp*y, xp])
        A.append([0, 0, 0, -x, -y, -1, yp*x, yp*y, yp])

    A = np.array(A)

    _, _, Vt = np.linalg.svd(A)

    H = Vt[-1].reshape(3, 3)

    return H / H[2, 2]

# Read image
img = cv2.imread('e11.jpg')

# Source and destination points
src = np.float32([[56,65], [368,52], [28,387], [389,390]])
dst = np.float32([[0,0], [300,0], [0,300], [300,300]])

# Compute Homography using DLT
H = dlt(src, dst)

# Apply transformation
result = cv2.warpPerspective(img, H, (300,300))

print("Original Image")
cv2_imshow(img)

print("DLT Result")
cv2_imshow(result)
```

```python
from google.colab import files
uploaded = files.upload()
import cv2
from google.colab.patches import cv2_imshow

# Read images
template = cv2.imread('watch.jpg', 0)
img = cv2.imread('scene.jpg')

# Check if images are loaded successfully
if template is None:
    print("Error: Template image 'watch.jpg' not found!")
elif img is None:
    print("Error: Scene image 'scene.jpg' not found!")
else:
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Get dimensions
    h_template, w_template = template.shape
    h_gray, w_gray = gray.shape

    if h_template > h_gray or w_template > w_gray:
        print(f"Warning: Template image (watch.jpg) is larger than the scene image (scene.jpg).")
        print(f"Template size: ({w_template}, {h_template}), Scene size: ({w_gray}, {h_gray})")
        print("Resizing template to fit within the scene image for matching.")
        # Resize template to fit within the scene image, maintaining aspect ratio
        scale = min(h_gray / h_template, w_gray / w_template)
        new_w = int(w_template * scale)
        new_h = int(h_template * scale)
        template = cv2.resize(template, (new_w, new_h), interpolation=cv2.INTER_AREA)
        print(f"New template size: ({new_w}, {new_h})")

    # Template Matching
    result = cv2.matchTemplate(gray, template, cv2.TM_CCOEFF_NORMED)

    # Find best match
    min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(result)

    # Get template size (after potential resizing)
    h, w = template.shape

    # Draw rectangle
    top_left = max_loc
    bottom_right = (top_left[0] + w, top_left[1] + h)

    cv2.rectangle(img, top_left, bottom_right, (0,255,0), 2)

    print("Detected Watch")
    cv2_imshow(img)
```

```python
from google.colab import files
uploaded = files.upload()
import cv2
from google.colab.patches import cv2_imshow
from IPython.display import clear_output
import time

# Open video
cap = cv2.VideoCapture('video.mp4')

frames = []

# Read all frames
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    frames.append(frame)

cap.release()

# Play video in reverse
for frame in reversed(frames):
    cv2_imshow(frame)
    time.sleep(0.03)
    clear_output(wait=True)

print("Reverse Video Playback Completed")
```

```python
from google.colab import files
uploaded = files.upload()
import cv2
from google.colab.patches import cv2_imshow

# Read image
img = cv2.imread('e34.jpg')

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Load Haar Cascade
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    'haarcascade_frontalface_default.xml'
)

# Detect faces
faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5,
    minSize=(30,30)
)

# Draw rectangles
for (x, y, w, h) in faces:
    cv2.rectangle(img,
                  (x, y),
                  (x+w, y+h),
                  (0,255,0),
                  2)

print("Detected Faces")
cv2_imshow(img)

print("Number of Faces Detected:", len(faces))
```

```python
from google.colab import files
uploaded = files.upload()
import cv2
from google.colab.patches import cv2_imshow
from IPython.display import clear_output

cap = cv2.VideoCapture('video11.mp4')

fgbg = cv2.createBackgroundSubtractorMOG2()

while True:
    ret, frame = cap.read()

    if not ret:
        break

    mask = fgbg.apply(frame)

    contours, _ = cv2.findContours(
        mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    for cnt in contours:
        if cv2.contourArea(cnt) > 1000:

            x, y, w, h = cv2.boundingRect(cnt)

            cv2.rectangle(
                frame,
                (x, y),
                (x+w, y+h),
                (0, 255, 0),
                2
            )

    cv2_imshow(frame)
    clear_output(wait=True)

cap.release()

print("Vehicle Detection Completed")
```
